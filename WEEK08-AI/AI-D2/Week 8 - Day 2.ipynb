Download the Planet eBook PDF

```python
import os
import urllib.request

pdf_url = "https://www.planetebook.com/free-ebooks/crime-and-punishment.pdf"
pdf_path = "crime_and_punishment.pdf"

# Download the PDF locally
if not os.path.exists(pdf_path):
    print("Downloading Crime and Punishment PDF...")
    urllib.request.urlretrieve(pdf_url, pdf_path)
    print("Download complete!")
else:
    print("File already exists.")
```

 Load the PDF Document

```python
from langchain_community.document_loaders import PyPDFLoader

# Load PDF pages
print("Loading PDF pages...")
loader = PyPDFLoader(pdf_path)
pages = loader.load()

print(f"Total pages loaded: {len(pages)}")
```

Chunking Experiment (Optimization Variable)

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Small chunk size setup
splitter_small = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks_small = splitter_small.split_documents(pages)

# Large chunk size setup
splitter_large = RecursiveCharacterTextSplitter(chunk_size=1200, chunk_overlap=150)
chunks_large = splitter_large.split_documents(pages)

print(f"Small chunks created: {len(chunks_small)}")
print(f"Large chunks created: {len(chunks_large)}")
```

Load Open-Source HuggingFace Embedding Model

```python
!pip install sentence-transformers
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

# Using open-source Hugging Face embeddings (no API key required)
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
print("Embedding model loaded successfully.")
```

Create and Store in Vector Database (ChromaDB)

```python
from langchain_community.vectorstores import Chroma

# Create small chunks database
db_small = Chroma.from_documents(
    documents=chunks_small,
    embedding=embedding_model,
    persist_directory="./chroma_db_small"
)

# Create large chunks database
db_large = Chroma.from_documents(
    documents=chunks_large,
    embedding=embedding_model,
    persist_directory="./chroma_db_large"
)

print("Vector stores indexed and saved successfully!")
```

Test Retrieval & Compare Findings

```python
# Test query for effective retrieval
test_query = "Why did Raskolnikov decide to kill the old woman?"

print("=== RETRIEVAL RESULTS (Small Chunks: 500) ===")
results_small = db_small.similarity_search(test_query, k=2)
for i, doc in enumerate(results_small, 1):
    print(f"\n[Result {i} | Page {doc.metadata.get('page', 'N/A')}]")
    print(doc.page_content.strip())

print("\n" + "="*50 + "\n")

print("=== RETRIEVAL RESULTS (Large Chunks: 1200) ===")
results_large = db_large.similarity_search(test_query, k=2)
for i, doc in enumerate(results_large, 1):
    print(f"\n[Result {i} | Page {doc.metadata.get('page', 'N/A')}]")
    print(doc.page_content.strip())
```
