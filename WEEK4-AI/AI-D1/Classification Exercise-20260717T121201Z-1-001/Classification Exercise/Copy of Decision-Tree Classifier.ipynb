#Decision Tree Classifier Lab


#Load essential libraries

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn import tree
from sklearn.model_selection import train_test_split
```

#Import the dataset

```python
df = pd.read_csv("Copy of titanic.csv")
df
```

##Drop meaningless information

```python
df = df.drop(columns = ['PassengerId','Name','SibSp', 'Parch', 'Ticket', 'Cabin', 'Embarked'], axis = 'columns')
df
```

##Data Preprocessing

```python
import math
mean_age = df.Age.mean()
mean_age = math.floor(mean_age)
print(mean_age)
```

```python
df['Age'] = df.Age.fillna(mean_age)
df
```

##Label Encode the non numeric features

```python
LE_sex = LabelEncoder()
df['Sex'] = LE_sex.fit_transform(df['Sex'])

df
```

## Define Feature Matrix (X) and Target Vector (y)


```python
y = df['Survived']
X = df.drop('Survived', axis = 'columns')
X
```

## Train-Test Split and Model Initialization


```python
model = tree.DecisionTreeClassifier()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.4)
```

##Train the model

```python
model.fit(X_train, y_train)

model.score(X_test, y_test)
```

##Predict results

```python
model.predict(X_test)
```

##Try decision tree classifer on the same target column but this time use more feature columns than before.

```python
#Code here

df2 = pd.read_csv("Copy of titanic.csv")

## Drop truly useless columns 
df2 = df2.drop(columns=['PassengerId', 'Name', 'Ticket', 'Cabin'], axis='columns')

## Handle missing Age
mean_age2 = math.floor(df2.Age.mean())
df2['Age'] = df2.Age.fillna(mean_age2)

## Handle missing Embarked (fill with the most common port)
df2['Embarked'] = df2['Embarked'].fillna(df2['Embarked'].mode()[0])

## Label Encode Sex and Embarked (both are text columns)
LE_sex2 = LabelEncoder()
df2['Sex'] = LE_sex2.fit_transform(df2['Sex'])

LE_embarked = LabelEncoder()
df2['Embarked'] = LE_embarked.fit_transform(df2['Embarked'])

## Define X and y with MORE features this time
y2 = df2['Survived']
X2 = df2.drop('Survived', axis='columns')  

## Train-test split
model2 = tree.DecisionTreeClassifier()
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.4)

## Train
model2.fit(X2_train, y2_train)

## Score — compare this to your first model's score!
print("Accuracy with more features:", model2.score(X2_test, y2_test))

## Predict
print("Predictions:", model2.predict(X2_test)) 
print("Actual:", y2_test)                      
```
