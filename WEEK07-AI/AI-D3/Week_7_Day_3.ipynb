## 2. Quick-start with Pipelines

The `pipeline` API wraps tokenization, model loading, and inference in one object.

### 2.1 Sentiment Analysis

```python
from transformers import pipeline

# 1. Load a sentiment-analysis pipeline (defaults to a small model)
sentiment = pipeline("sentiment-analysis")

# 2. Run inference
examples = [
    "Hugging Face makes NLP super accessible!",
    "I dislike bugs in my code..."
]
results = sentiment(examples)
for text, res in zip(examples, results):
    print(f"{text!r:50} → label={res['label']}, score={res['score']:.3f}")
```

### 2.2 Text Generation

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "Once upon a time"
out = generator(prompt, max_length=30, num_return_sequences=1)
print(out[0]["generated_text"])
```

### 2.3 Question Answering

```python
from transformers import pipeline

qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
context = (
    "Hugging Face is an AI company with the mission to democratize good machine learning. "
    "The Transformers library provides thousands of pretrained models in 100+ languages."
)

answer = qa({
    "question": "What is the mission of Hugging Face?",
    "context": context
})
print(answer["answer"])
```

## 1. Selecting a Pre-trained Model

For this assignment, I selected the `distilgpt2` model from Hugging Face.

### Why this model?

- It is designed for text generation.
- It is smaller and faster than the original GPT-2.
- It can generate text from a given prompt.
- It works directly with the Hugging Face `text-generation` pipeline.

````python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
````

````python
prompt = "Artificial intelligence is"

result = generator(
    prompt,
    max_length=40,
    num_return_sequences=1
)

print(result[0]["generated_text"])
````

## 2. Demonstration on Three Examples

The pipeline will be tested using three different prompts.

````python
examples = [
    "Artificial intelligence is",
    "The future of technology will",
    "Machine learning helps us"
]

for prompt in examples:
    result = generator(
        prompt,
        max_length=40,
        num_return_sequences=1
    )

    print("Prompt:", prompt)
    print("Generated:", result[0]["generated_text"])
    print("-" * 80)
````

## 3. Experiment 1 — Effect of max_length

`max_length` controls the maximum number of tokens in the generated output.

We will compare a short output with a longer output.

````python
prompt = "Machine learning is changing the world"

# Short output
short_result = generator(
    prompt,
    max_length=30,
    num_return_sequences=1
)

# Longer output
long_result = generator(
    prompt,
    max_length=70,
    num_return_sequences=1
)

print("MAX_LENGTH = 30")
print(short_result[0]["generated_text"])

print("\n" + "=" * 80)

print("MAX_LENGTH = 70")
print(long_result[0]["generated_text"])
````

### Observation

When `max_length` is increased from 30 to 70, the model is allowed to
generate a longer response. Therefore, the second output is generally
longer than the first output.

## 4. Experiment 2 — Effect of Temperature

The `temperature` parameter controls the randomness of text generation.

- Lower temperature produces more predictable text.
- Higher temperature produces more diverse and creative text.

````python
prompt = "The future of artificial intelligence"

# Low temperature
low_temp = generator(
    prompt,
    max_length=50,
    temperature=0.2,
    do_sample=True,
    num_return_sequences=1
)

# High temperature
high_temp = generator(
    prompt,
    max_length=50,
    temperature=1.0,
    do_sample=True,
    num_return_sequences=1
)

print("TEMPERATURE = 0.2")
print(low_temp[0]["generated_text"])

print("\n" + "=" * 80)

print("TEMPERATURE = 1.0")
print(high_temp[0]["generated_text"])
````

### Observation

At a lower temperature such as 0.2, the generated text tends to be
more predictable and focused.

At a higher temperature such as 1.0, the model produces more varied
and creative text.

Therefore, increasing temperature generally increases randomness
during text generation.

## 5. Experiment 3 — Effect of top_k

The `top_k` parameter limits the model to selecting the next token
from the top K most likely choices.

A smaller `top_k` generally restricts the choices, while a larger
`top_k` allows more possible choices.

````python
prompt = "Technology is changing"

# Small top_k
small_top_k = generator(
    prompt,
    max_length=50,
    top_k=10,
    do_sample=True,
    num_return_sequences=1
)

# Larger top_k
large_top_k = generator(
    prompt,
    max_length=50,
    top_k=100,
    do_sample=True,
    num_return_sequences=1
)

print("TOP_K = 10")
print(small_top_k[0]["generated_text"])

print("\n" + "=" * 80)

print("TOP_K = 100")
print(large_top_k[0]["generated_text"])
````

### Observation

With `top_k=10`, the model selects the next words from a smaller set
of likely choices, which can make the output more focused.

With `top_k=100`, the model has a larger set of possible words to
choose from, which can make the generated text more diverse.

The exact output may vary because text generation involves sampling.

## 6. Parameter Comparison

| Parameter | Setting 1 | Setting 2 | Effect |
|---|---:|---:|---|
| max_length | 30 | 70 | Controls output length |
| temperature | 0.2 | 1.0 | Controls randomness |
| top_k | 10 | 100 | Controls the number of likely token choices |

### Overall Conclusion

The Hugging Face text-generation pipeline makes it easy to use a
pre-trained language model without training a model from scratch.

The experiments demonstrated that different parameters can change
the generated output:

- `max_length` controls how long the output can be.
- `temperature` controls the randomness of generation.
- `top_k` controls how many likely token choices are considered.

Therefore, model parameters can significantly affect the style and
length of generated text.
